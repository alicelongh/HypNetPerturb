```python
def load_connectivity_from_mat(path, imaging=None, key="temp", freq_index=2):
    """
    Load a connectivity matrix from a .mat file, handling DWI and fMRI formats.

    Parameters
    ----------
    path : str
        Path to the .mat file.

    imaging : {"DWI", "fMRI", None}, default=None
        Imaging modality:
        - "DWI": reads mat[key]
        - "fMRI": reads mat[key][0][0][0][:, :, freq_index]
        - None: reads mat[key] directly

    key : str, default="temp"
        Variable name inside the .mat file.

    freq_index : int, default=2
        Frequency index to select for fMRI data.

    Returns
    -------
    A : np.ndarray
        Loaded adjacency/connectivity matrix.
    """
    mat = scio.loadmat(path)

    if imaging == "DWI":
        A = mat[key]

    elif imaging == "fMRI":
        A = mat[key][0][0][0][:, :, freq_index]

    else:
        A = mat[key]

    return A
```
